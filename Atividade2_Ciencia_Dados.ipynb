Nesta atividade, será feita uma análise das características financeiras do dataset de clientes de cartão de crédito. Serão analisadas as colunas de valor da fatura e valor de pagamento, utilizando pandas, NumPy e matplotlib.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('C:\Arquivos_trabalhos\default_of_credit_card_clients__courseware_version_1_21_19.xls')

df.head()
```

## Exercício 1

Criação das listas com os nomes das características financeiras.

```python
bill_amt_features = [
    'BILL_AMT1',
    'BILL_AMT2',
    'BILL_AMT3',
    'BILL_AMT4',
    'BILL_AMT5',
    'BILL_AMT6'
]

pay_amt_features = [
    'PAY_AMT1',
    'PAY_AMT2',
    'PAY_AMT3',
    'PAY_AMT4',
    'PAY_AMT5',
    'PAY_AMT6'
]

print(bill_amt_features)
print(pay_amt_features)
```

## Exercício 2

Análise estatística das características de valor da fatura.

```python
df[bill_amt_features].describe()
```

Reflexão: Sim, os resultados fazem sentido. Algumas colunas de fatura apresentam valores negativos, que podem indicar crédito para o cliente, estorno ou pagamento maior que o valor devido. Além disso, os valores máximos são bem maiores que a maior parte dos dados, mostrando que existem clientes com faturas muito altas em comparação aos demais. Isso indica a presença de valores extremos no dataset.

## Exercício 3

Histogramas das características de valor da fatura.

```python
df[bill_amt_features].hist(
    bins=20,
    layout=(2, 3),
    figsize=(12, 8)
)
```

Os histogramas mostram que a maioria dos valores de fatura está concentrada em faixas menores. Também aparecem alguns valores muito altos, indicando clientes com saldos maiores nas faturas.

## Exercício 4

Resumo estatístico das características de valor do pagamento.

```python
df[pay_amt_features].describe()
```

Reflexão: Os resultados fazem sentido. Os pagamentos têm valor mínimo igual a 0, pois alguns clientes podem não ter realizado pagamento em certos meses. Também é possível perceber que existem poucos pagamentos muito altos, o que aumenta a média e mostra que os dados não estão distribuídos de forma equilibrada.

## Exercício 5

Histogramas das características de pagamento.

```python
df[pay_amt_features].hist(
    bins=20,
    layout=(2, 3),
    figsize=(12, 8),
    xrot=45
)
```

Os histogramas mostram uma grande concentração de valores próximos de zero. Isso indica que muitos clientes fizeram pagamentos baixos ou nenhum pagamento em alguns meses. Como existem valores muito altos, os histogramas ficam bastante concentrados no início da escala.


## Exercício 6

Verificação dos pagamentos iguais a zero usando máscara booleana.

```python
pay_zero_mask = df[pay_amt_features] == 0

pay_zero_mask.sum()
```

Reflexão: A quantidade de pagamentos iguais a zero confirma o que foi observado nos histogramas. Como os gráficos mostraram grande concentração perto de zero, faz sentido encontrar muitos pagamentos exatamente iguais a zero.

## Exercício 7

Histogramas das transformações logarítmicas dos pagamentos diferentes de zero.


```python
non_zero_payments = df[pay_amt_features][df[pay_amt_features] != 0]

log_payments = non_zero_payments.apply(np.log10)

log_payments.hist(
    bins=20,
    layout=(2, 3),
    figsize=(12, 8),
    xrot=45
)
```

Reflexão: Ao ignorar os pagamentos iguais a zero e aplicar a transformação logarítmica, os dados ficam mais fáceis de visualizar. Isso ocorre porque os valores de pagamento variam muito, e o logaritmo reduz o impacto dos valores extremos.
